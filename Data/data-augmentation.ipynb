```python
import pandas as pd

def read_excel_to_dict(file_path):
    class_dict = {}

    # Read Excel file
    df = pd.read_excel(file_path, header=None, dtype={2: str})  # Ensure column 2 is read as string
    df.fillna("", inplace=True)  # Replace NaN with empty string

    for _, row in df.iterrows():
        class_name, super_label = row.iloc[1].strip(), row.iloc[2].strip()

        if super_label == "":  # It's a super class
            class_dict[class_name] = []
        else:  # It's a subclass
            if super_label not in class_dict:
                class_dict[super_label] = []
            class_dict[super_label].append(class_name)

    return class_dict

if __name__ == "__main__":
    file_path = "/kaggle/input/video-data/Book1.xlsx"  # Change this to your actual file path
    class_hierarchy = read_excel_to_dict(file_path)

    for super_label, classes in class_hierarchy.items():
      print(f"Super Class {super_label}: {classes}")

    print(class_hierarchy)
```

```python
def replace_superclass_labels(class_dict):
    superclass_labels = {
        "1": "Eating and Drinking",
        "2": "Working (Productive)",
        "3": "Violence",
        "4": "Physical Exercise",
        "5": "Playing Instruments",
        "6": "Smoke",
        "7": "Cleaning",
        "8": "Celebration",
        "9": "Repair and Maintenance",
        "10": "Theft",
        "11": "Fidget",
        "12": "Using Phone",
        "13": "Recreation Activities",
        "14": "Distraction (Unproductive)",
        "15": "Medical Emergency",
        "16": "Fire Emergency"
    }
    sorted_keys = sorted(superclass_labels.keys(), key=int)  # Maintain numerical order
    updated_dict = {}

    for key, value in class_dict.items():
        new_key = superclass_labels.get(key, key)  # Replace if mapping exists, else keep original
        updated_dict[new_key] = value if value else [new_key]

    return updated_dict

if __name__ == "__main__":
    updated_hierarchy = replace_superclass_labels(class_hierarchy)
    print(updated_hierarchy)
```

```python
import os

def count_videos(directory, class_dict):
    result = {}

    for group, classes in class_dict.items():
        group_total = 0
        result[group] = {}

        for class_name in classes:
            class_path = os.path.join(directory, class_name)

            if os.path.isdir(class_path):
                video_count = len([f for f in os.listdir(class_path) if f.endswith(('.mp4', '.avi', '.mov', '.mkv'))])
            else:
                video_count = 0

            result[group][class_name] = video_count
            group_total += video_count

        result[group]['Total'] = group_total

    return result

def print_result(result):
    for group, classes in result.items():
        print(f"{group}:")
        for class_name, count in classes.items():
            print(f"  {class_name}: {count} videos")
        print()

# Set your directory path here
directory = "/kaggle/input/video-data/train/train"
result = count_videos(directory, updated_hierarchy)
print_result(result)
```

```python
import os
import random
import pandas as pd
import cv2
import numpy as np
from moviepy.editor import VideoFileClip, concatenate_videoclips
from datetime import datetime

def load_class_labels(class_labels_file):
    class_labels_df = pd.read_excel(class_labels_file, engine='openpyxl')  # Updated to read Excel file
    class_labels = dict(zip(class_labels_df.iloc[:, 1], class_labels_df.iloc[:, 0]))  # {Class Name: Class Number}
    return class_labels

def get_video_paths(dataset_path, selected_classes, class_labels):
    class_videos = {}
    video_usage = {}
    
    for class_name in selected_classes:
        class_folder = os.path.join(dataset_path, class_name)
        if not os.path.isdir(class_folder):
            print(f"Warning: Class folder {class_name} not found.")
            continue
        
        videos = [os.path.join(class_folder, vid) for vid in os.listdir(class_folder) if vid.endswith(('.mp4', '.avi', '.mov'))]
        if videos:
            class_videos[class_name] = videos
            video_usage[class_name] = set(videos)
    
    return class_videos, video_usage

def create_augmented_videos(dataset_path, selected_classes, num_videos, output_dir, class_labels_file, output_folder_name):
    class_labels = load_class_labels(class_labels_file)
    class_videos, video_usage = get_video_paths(dataset_path, selected_classes, class_labels)
    
    if not class_videos:
        print("No valid classes found for augmentation!")
        return
    
    output_run_dir = os.path.join(output_dir, output_folder_name)
    os.makedirs(output_run_dir, exist_ok=True)
    
    print(f"Output files will be saved in: {output_run_dir}")
    
    metadata_list = []
    video_usage_stats = {class_name: 0 for class_name in class_videos.keys()}
    
    for i in range(num_videos):
        selected_videos = []
        metadata = []
        start_time = 0.0
        
        while start_time < 10.0:
            class_name = random.choice(list(class_videos.keys()))
            
            if not video_usage[class_name]:
                video_usage[class_name] = set(class_videos[class_name])
            
            video_path = video_usage[class_name].pop()
            
            try:
                video_clip = VideoFileClip(video_path)
                max_start = max(0, video_clip.duration - 2)  # Ensure random start time within video duration
                start_clip = random.uniform(0, max_start)
                clip_duration = min(2, video_clip.duration - start_clip, 10.0 - start_time)  # Ensure correct duration
                clip = video_clip.subclip(start_clip, start_clip + clip_duration)
                selected_videos.append(clip)
                
                class_label = class_labels.get(class_name, -1)
                metadata.append([f"aug_video_{i+1}.mp4", f"{start_time:.1f}-{start_time+clip_duration:.1f}", 
                                 class_label, class_name, os.path.basename(video_path)])
                start_time += clip_duration
                video_usage_stats[class_name] += 1
            except Exception as e:
                print(f"Skipping video {video_path} due to error: {e}")
                continue
        
        if selected_videos:
            final_video = concatenate_videoclips(selected_videos, method='compose')
            output_video_path = os.path.join(output_run_dir, f"aug_video_{i+1}.mp4")
            final_video.write_videofile(output_video_path, codec='libx264', fps=30)
            metadata_list.extend(metadata)
    
    metadata_df = pd.DataFrame(metadata_list, columns=["Augmented Video", "Duration", "Class Number Label", "Class Name", "Video File Name"])
    metadata_df.to_csv(os.path.join(output_run_dir, "aug_metadata.csv"), index=False)
    
    # Save video usage statistics as CSV
    usage_data = [[class_labels.get(class_name, -1), class_name, len(class_videos[class_name]), video_usage_stats[class_name]] for class_name in class_videos.keys()]
    usage_df = pd.DataFrame(usage_data, columns=["Class Number Label", "Class Name", "Total Videos", "Videos Used for Augmentation"])
    usage_df.to_csv(os.path.join(output_run_dir, "video_usage_stats.csv"), index=False)
    
    print(f"All augmented videos, metadata, and video statistics saved in {output_run_dir}")
    return output_run_dir

if __name__ == "__main__":
    dataset_path = '/kaggle/input/video-data/train/train'
    output_dir = '/kaggle/working/'
    class_labels_file = '/kaggle/input/video-data/Book1.xlsx'  # Updated prompt for Excel file
```
